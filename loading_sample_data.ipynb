# Loading Sample Data
Sample data is available for free from CarAPI for 2015-2020. This data is availble as CSV files, so the following notebook will load the CSV files, join them, filter to the appropriate columns, and export the full sample data set that we will use to create the tool prototype. In the final iteration of this project, we will use data directly from the API.

```python
import os
from pathlib import Path
import codecarbon.emissions_tracker as et

et.count_physical_cpus = lambda: 1

from codecarbon import EmissionsTracker

LOG_DIR = Path("data/carbon_logs")
LOG_DIR.mkdir(parents=True, exist_ok=True)

tracker = EmissionsTracker(
    project_name="sample_data_load",
    output_dir=str(LOG_DIR),
    output_file="sample_data_load.csv",
)
tracker.start()
```

```python
# load all of the csvs
import pandas as pd

#trims
t_df = pd.read_csv('data/data_2015_2020/trims-sample.csv')
#removing spaces and uppercase letters
t_df.columns = t_df.columns.str.strip().str.lower().str.replace(' ', '_')
#selecting relevant columns
t_df = t_df[["make_id", "model_id", "submodel_id","trim_id","model_year","make_name","model_name",
    "submodel_name","trim_name","trim_description","trim_msrp","trim_invoice"]]
#changing column names to match eventaul API calls
t_df = t_df.rename(columns={'model_year': 'year', 'make_name': 'make', 'model_name' : 'model', 'submodel_name' : 'submodel',
'trim_name': 'trim','trim_description': 'description', 'trim_msrp': 'msrp', 'trim_invoice': 'invoice'})

#bodies
b_df = pd.read_csv('data/data_2015_2020/bodies-sample.csv')
#removing spaces and uppercase letters
b_df.columns = b_df.columns.str.strip().str.lower().str.replace(' ', '_')
#selecting relevant columns
b_df = b_df[["make_id", "model_id", "submodel_id","trim_id","body_type"]]

#engines
e_df = pd.read_csv('data/data_2015_2020/engines-sample.csv')
#removing spaces and uppercase letters
e_df.columns = e_df.columns.str.strip().str.lower().str.replace(' ', '_')
#selecting relevant columns
e_df = e_df[["make_id", "model_id", "submodel_id","trim_id","engine_type", "engine_fuel_type"]]
#changing column names to match eventaul API calls
e_df = e_df.rename(columns={'engine_fuel_type': 'fuel_type'})

#milealges
m_df = pd.read_csv('data/data_2015_2020/mileages-sample.csv')
#removing spaces and uppercase letters
m_df.columns = m_df.columns.str.strip().str.lower().str.replace(' ', '_')
#selecting relevant columns
m_df = m_df[["make_id", "model_id", "submodel_id","trim_id","mileage_fuel_tank_capacity", "mileage_combined_mpg", "mileage_epa_city_mpg", 
"mileage_epa_highway_mpg","mileage_range_city", "mileage_range_highway", "mileage_battery_capacity_electric",
            "mileage_epa_time_to_charge_hr_240v_electric", "mileage_epa_kwh_100_mi_electric",
            "mileage_range_electric", "mileage_epa_highway_mpg_electric", "mileage_epa_city_mpg_electric",
            "mileage_epa_combined_mpg_electric"]]
#changing column names to match eventaul API calls
m_df.columns = m_df.columns.str.replace('mileage_', '')
```

```python
full_df = (
    t_df.merge(b_df, on=["make_id", "model_id", "submodel_id", "trim_id"], how="left")
    .merge(e_df, on=["make_id", "model_id", "submodel_id", "trim_id"], how="left")
    .merge(m_df, on=["make_id", "model_id", "submodel_id", "trim_id"], how="left")
)
```

```python
full_df.to_csv( 'data/data_2015_2020/full_dataset.csv',index=False)
```
